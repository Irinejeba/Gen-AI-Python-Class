```python
import re

# File path configuration
file_path = "C:\\Users\\manju\\Desktop\\vengat.txt"

# Read the file content once to use across different examples
try:
    with open(file_path, "r", encoding="utf-8") as myfile:
        mydata = myfile.read()
except FileNotFoundError:
    print(f"Error: The file at {file_path} was not found.")
    print("Please verify your path or place a test file there to run the script.")
    mydata = ""

if mydata:
    # --- 1. re.findall() - Case Insensitive matching for "and" ---
    print("--- 1. re.findall() [Aa][n][d] ---")
    mylist_and = re.findall(r"[Aa][n][d]", mydata)
    print(mylist_and)
    print()


    # --- 2. re.findall() - Extracting Dates (DD-MM-YYYY formats) ---
    print("--- 2. re.findall() Extracting Dates ---")
    mylist_dates = re.findall(r"\d{2}[-/.]\d{2}[-/.]\d{4}", mydata)
    print(mylist_dates)
    
    # Print each discovered date individually
    for date in mylist_dates:
        print(date)
    print()


    # --- 3. re.split() - Splitting string by space ---
    print("--- 3. re.split() Space Separator ---")
    mylist_split = re.split(r" ", mydata)
    for word in mylist_split:
        if word.strip():  # Filters out trailing/unnecessary newlines for display
            print(word)
    print()


    # --- 4. re.sub() - Replacing spaces with newlines ---
    print("--- 4. re.sub() Replace Spaces ---")
    replaced_text = re.sub(r" ", '\n', mydata)
    print(replaced_text)
    print()


    # --- 5. re.search() - Finding first occurrence match object ---
    print("--- 5. re.search() First Match ---")
    first_match = re.search(r"\d{2}[-/.]\d{2}[-/.]\d{4}", mydata)
    print(first_match)
```

```python
import json
import os

def generate_json_log(start_path, output_json="directory_map.json"):
    print(f"[*] Compiling document structure map layout into: {output_json}")
    
    directory_data = []
    
    # Secure walk loop through active file structures
    for root, dirs, files in os.walk(start_path):
        folder_node = {
            "folder_path": root,
            "subdirectories": dirs,
            "files": files
        }
        directory_data.append(folder_node)
        
    try:
        with open(output_json, "w", encoding="utf-8") as json_file:
            # Indent layout flag guarantees scannable configuration outputs
            json.dump(directory_data, json_file, indent=4, ensure_ascii=False)
            
        print(f"[+] JSON layout database exported safely: {os.path.abspath(output_json)}")
    except Exception as e:
        print(f"[-] Unhandled pipeline exception context: {e}")

if __name__ == "__main__":
    target_directory = r"C:\Users\manju\Documents"
    generate_json_log(target_directory)
```
