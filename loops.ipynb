```python
# ==========================================
# PART 1: WHILE LOOPS & CONTROL FLOW
# ==========================================

# 1. Basic iterative forward loop (1 to 10)
a = 1  # Initialization
while a <= 10:  # Condition boundary
    print(a, end=" ")
    a += 1  # Incrementation step
print("\nLoop Over")

# 2. Stepped skip forward loop (0 to 100 by tens)
a = 0
while a <= 100:
    print(a, end=" ")
    a += 10
print("\nLoop Over")

# 3. Reverse countdown loop (100 down to 0)
a = 100
while a >= 0:
    print(a, end=" ")
    a -= 10
print("\nLoop Over")

# 4. Breaking a loop prematurely
a = 1
while a > 0:
    if a == 5:
        print(f"\nFound target value {a}, breaking out!")
        break
    print(a, end=" ")
    a += 1

# 5. Using 'continue' to skip iterations (Skipping multiples of 3)
a = 1
while a <= 20:
    if a % 3 == 0:
        print(f"({a})", end=" ")  # Marks the element being skipped
        a += 1
        continue  # Instantly skips back up to the loop boundary line
    a += 1
print("\nLoop over")


# ==========================================
# PART 2: NESTED WHILE LOOPS (GRID PATTERNS)
# ==========================================

# Static grid print task (5 rows x 5 columns)
row_idx = 1
while row_idx <= 5:
    col_idx = 1
    while col_idx <= 5:
        print(col_idx, end="")
        col_idx += 1
    row_idx += 1
    print()  # Moves cursor down to the next row line

# Dynamic User-Defined Grid Task Solution
print("\n--- User Defined Matrix Generator ---")
user_rows = int(input("Enter number of rows: "))
user_cols = int(input("Enter number of columns: "))

r = 1
while r <= user_rows:
    c = 1
    while c <= user_cols:
        print(c, end=" ")
        c += 1
    print()
    r += 1


# ==========================================
# PART 3: FOR LOOPS & THE RANGE() FUNCTION
# ==========================================

# 1. Iterating through a string value
mystr = "Hello world"
for char in mystr:
    print(char, end="")
print()

# 2. Iterating through item elements inside a list array
mylist = [10, 20.55, "hello", True, 20.55, "hello"]
for element in mylist:
    print(element, end=", ")
print()

# 3. range(stop) -> Generates numbers from 0 up to (but excluding) stop value
for x in range(11):  # 0 to 10
    print(x, end=" ")
print()

# 4. range(start, stop) -> Custom start index constraints
for x in range(5, 11):  # 5 to 10
    print(x, end=" ")
print()

# 5. range(start, stop, step) -> Stepping / Jump iterations forward
for x in range(100, 1001, 100):  # 100 to 1000
    print(x, end=" ")
print()

# 6. Reverse range loops using negative steps
for x in range(100, 0, -10):  # 100 down to 10
    print(x, end=" ")
print()


# ==========================================
# PART 4: FOR-ELSE BLOCKS & NESTED FOR LOOPS
# ==========================================
# Rule: The 'else' block triggers ONLY if the loop finishes without hitting a 'break'

# Example A: Unbroken loop (else executes)
for x in "Hello world":
    print(x, end="")
else:
    print("\n-> The loop above completed successfully!")

# Example B: Broken loop parsing a space character (else is skipped)
for x in "Hello world":
    if x == ' ':
        break
    print(x, end="")
else:
    print("\nThis line will not print because of the space break.")
print("\n-> Space encountered. Loop broke prematurely.")

# Example C: Linear Search verification via for-else
search_list = [2, 6, 9, 10, 15, 11, 16, 97]
for num in search_list:
    if num % 5 == 0:
        print(f"Target found: {num} is Divisible By 5")
        break
else:
    print("No numbers divisible by 5 were discovered in the dataset.")


# ==========================================
# PART 5: NESTED FOR LOOPS
# ==========================================

# Generating a grid pattern using range loops
print("\nNested Range Grid:")
for row in range(5):
    for col in range(1, 6):
        print(col, end="")
    print()

# Permutation combinations matching matrix mapping elements
print("\nNested Collection Combinations:")
colors = ["Yellow", "Green", "Red"]
entities = ["Cat", "Dog", "Car", "House"]

for color in colors:
    print("-" * 25)
    for entity in entities:
        print(f"{color} {entity}")
print("-" * 25)
print("Loop Over")
```
