#### 3. Create a class Rectangle with attributes length and width. 
####    Add methods to calculate area and perimeter.

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width

    def calculate_perimeter(self):
        return 2 * (self.length + self.width)

print("--- Question 3 ---")
rect = Rectangle(10, 5)
print(f"Rectangle Area: {rect.calculate_area()}")
print(f"Rectangle Perimeter: {rect.calculate_perimeter()}\n")
```

#### 5. Define a class Circle with a class variable pi = 3.14. 
####      Add a method to calculate the area.

```python
class Circle:
    pi = 3.14  # Class variable

    def __init__(self, radius):
        self.radius = radius  # Instance variable

    def calculate_area(self):
        return Circle.pi * (self.radius ** 2)

print("--- Question 5 ---")
my_circle = Circle(7)
print(f"Circle Area (Radius 7): {my_circle.calculate_area()}\n")
```

#### 6. Create a class BankAccount with attributes account_number, name, and balance. 
####      Add methods to deposit, withdraw, and display balance.

```python
class BankAccount:
    def __init__(self, account_number, name, balance=0.0):
        self.account_number = account_number
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited: ${amount}. New Balance: ${self.balance}")
        else:
            print("Invalid deposit amount.")

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew: ${amount}. Remaining Balance: ${self.balance}")
        else:
            print("Insufficient funds or invalid amount.")

    def display_balance(self):
        print(f"Account: {self.account_number} | Owner: {self.name} | Current Balance: ${self.balance}")

print("--- Question 6 ---")
account = BankAccount("BA1001", "Rohan", 500)
account.display_balance()
account.deposit(200)
account.withdraw(150)
print()
```

#### 7. Write a class Employee with attributes emp_id, name, and salary. 
####      Add a method to give a 10% bonus to employees.

```python
class Employee:
    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary

    def give_bonus(self):
        bonus = self.salary * 0.10
        self.salary += bonus
        print(f"10% Bonus applied! Added: ${bonus}. New Salary: ${self.salary}")

print("--- Question 7 ---")
emp = Employee("E401", "Vikram", 60000)
print(f"Original Salary: ${emp.salary}")
emp.give_bonus()
print()
```

#### 8. Implement a class Library with methods to add_book(title), 
####      remove_book(title), and display_books().

```python
class Library:
    def __init__(self):
        self.books = []  # List to store book titles

    def add_book(self, title):
        self.books.append(title)
        print(f"Added: '{title}' to the library.")

    def remove_book(self, title):
        if title in self.books:
            self.books.remove(title)
            print(f"Removed: '{title}' from the library.")
        else:
            print(f"Error: '{title}' not found in the library.")

    def display_books(self):
        if not self.books:
            print("The library is currently empty.")
        else:
            print("Library Catalog:")
            for book in self.books:
                print(f" - {book}")

print("--- Question 8 ---")
my_library = Library()
my_library.add_book("Python Crash Course")
my_library.add_book("Clean Code")
my_library.display_books()
my_library.remove_book("Clean Code")
my_library.display_books()
print()
```

#### 9. Create a class Calculator with methods add, subtract, multiply, and divide. 
####      Demonstrate with objects.

```python
class Calculator:
    def add(self, a, b):
        return a + b

    def subtract(self, a, b):
        return a - b

    def multiply(self, a, b):
        return a * b

    def divide(self, a, b):
        if b == 0:
            return "Error: Cannot divide by zero."
        return a / b

print("--- Question 9 ---")
calc = Calculator()
print(f"10 + 5 = {calc.add(10, 5)}")
print(f"10 - 5 = {calc.subtract(10, 5)}")
print(f"10 * 5 = {calc.multiply(10, 5)}")
print(f"10 / 0 = {calc.divide(10, 0)}")
print()
```

#### 10. Write a program to show the difference between instance variables 
####       and class variables with an example.

```python
class School:
    # Class Variable: Shared by all objects/instances of this class
    school_name = "Global High School"

    def __init__(self, student_name):
        # Instance Variable: Unique to each individual object/instance
        self.student_name = student_name

print("--- Question 10 ---")
s1 = School("Amit")
s2 = School("Pooja")

print("Initial Values:")
print(f"s1: Student = {s1.student_name}, School = {s1.school_name}")
print(f"s2: Student = {s2.student_name}, School = {s2.school_name}")

# Modifying the Class Variable using the Class name
School.school_name = "Excel Academy"

print("\nAfter changing the Class Variable (School.school_name):")
print(f"s1: Student = {s1.student_name}, School = {s1.school_name} (Updated!)")
print(f"s2: Student = {s2.student_name}, School = {s2.school_name} (Updated!)")

# Modifying an individual instance variable
s1.student_name = "Amit Kumar"

print("\nAfter changing an Instance Variable (s1.student_name):")
print(f"s1: Student = {s1.student_name} (Updated!)")
print(f"s2: Student = {s2.student_name} (Stays the same)")
```
