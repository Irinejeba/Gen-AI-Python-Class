### **Python Error Handling - Assessment Answers**

#### 1. What is the difference between syntax errors and runtime errors in Python?
*   **Syntax Errors:** Occur during the parsing stage *before* the code is executed. They happen when the code violates Python's structural rules (e.g., missing colons, mismatched parentheses, or incorrect indentation). The program will fail to start.

*   **Runtime Errors (Exceptions):** Occur while the program is actively executing. The syntax is perfectly valid, but an abnormal condition or invalid operation occurs during runtime (e.g., dividing by zero, trying to open a file that doesn't exist, or referencing an undefined variable).

#### 2. Explain the role of the try, except, else, and finally blocks.
*   **`try`:** Wraps the block of code that you suspect might raise an exception. Python monitors this block for errors.
*   **`except`:** Contains the code that executes *only if* a matching exception is raised inside the `try` block. It handles or mitigates the error.
*   **`else`:** Executes *only if* the code in the `try` block runs completely smoothly without raising any exceptions.
*   **`finally`:** Executes *always*, regardless of whether an exception was raised, caught, or completely unhandled. It is typically used for cleanup tasks like closing files or network connections.


#### 3. What happens if an exception is raised but not handled?
If an exception is raised and no matching `except` block handles it, the execution of the program **halts immediately**. Python then prints a detailed **traceback** to the standard error output, showing the line of code where the error occurred and the chain of function calls leading up to it.


### 4. What is the difference between Exception and BaseException in Python?
*   #**`BaseException`:** The absolute root class for all built-in exceptions. 
*   **`Exception`:** A direct subclass of `BaseException`. 
*   **Key Distinction:** `BaseException` includes system-level events and exit signals like `SystemExit`, `KeyboardInterrupt` (Ctrl+C), and `GeneratorExit`. `Exception` serves as the base class for all standard, application-level, and user-defined errors. You should almost always catch `Exception` instead of `BaseException` so you don't accidentally block the user from stopping the script.


#### 5. Explain the use of the raise keyword with an example.
The `raise` keyword is used to **explicitly force** an exception to occur. This is highly useful for input validation or forcing specific behavior when business rules are violated.

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ZeroDivisionError:
    print("Error: You cannot divide by zero!")
except ValueError:
    print("Error: Please enter a valid integer!")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

#### 6. How can you create a custom exception class in Python?
To create a custom exception, define a new class that inherits from the built-in **`Exception`** class (or one of its specific subclasses).

```python
class InsufficientFundsError(Exception):
    """Raised when a bank account balance drops below the withdrawal amount."""
    def __init__(self, balance, amount):
        super().__init__(f"Attempted to withdraw ${amount} but balance is only ${balance}.")
        self.balance = balance
        self.amount = amount

# Usage
try:
    raise InsufficientFundsError(50, 100)
except InsufficientFundsError as e:
    print(e)
```

#### 7. What is the purpose of the assert statement, and how does it differ from raising an exception?
*   **Purpose:** The `assert` statement evaluates a conditional expression. If the expression evaluates to `False`, it triggers an `AssertionError`. It is meant for internal debugging and sanity checks during development.
*   **Difference:** Unlike an explicit `raise` statement, assertions can be **globally disabled** at compilation. If you run Python with the optimization flag (`python -O script.py`), all `assert` statements are stripped out. Therefore, you should never use `assert` for production-critical workflows like user input validation.

#### 8. Can we use multiple except blocks in Python? Give an example.
Yes, a single `try` block can be followed by multiple `except` blocks to handle different types of exceptions distinctively. Python evaluates them sequentially from top to bottom and runs the *first matching* block.

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ZeroDivisionError:
    print("Error: You cannot divide by zero!")
except ValueError:
    print("Error: Please enter a valid integer!")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```


*Note: Always place the most specific exceptions at the top and broader exceptions (like generic `Exception`) at the bottom.*



#### 9. What happens if you use a finally block without an except block?
Using `try...finally` without an `except` block is **entirely valid syntax**. 
If an exception occurs inside the `try` block, Python will temporarily pause the error propagation, execute the code inside the `finally` block to ensure cleanup, and *then* immediately raise the unhandled exception afterward to halt the program.

##### 10. What is the difference between checked and unchecked exceptions? Does Python support checked exceptions?
*   **Difference:** *Checked exceptions* are checked at compile-time; the language forces the developer to either handle the exception inside a block or declare it in the function's signature (like in Java). *Unchecked exceptions* are evaluated at runtime and do not require strict declarations.
*   **Python Support:** **No**, Python does not support checked exceptions. All exceptions in Python are unchecked, relying purely on dynamic runtime execution and flexible error handling.
