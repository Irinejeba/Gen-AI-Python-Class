```python
# =====================================================================
# INTERVIEW PREPARATION: Palindrome Checking Algorithms
# =====================================================================

# ---------------------------------------------------------------------
# Approach 1: Simple Reverse (Works only for exact, lowercase matches)
# ---------------------------------------------------------------------
print("--- Approach 1: Simple Reverse ---")
mystr = "racecar"
revstr = mystr[::-1]  # Reverses the string using slice step -1

print("Original:", mystr)
print("Reversed:", revstr)

if mystr == revstr:
    print("Is Palindrome:", True)
else:
    print("Is Palindrome:", False)
print()


# ---------------------------------------------------------------------
# Approach 2: String Formatting using .replace() and .lower()
# (Fixes case insensitivity and simple spacing issues)
# ---------------------------------------------------------------------
print("--- Approach 2: Lowercase and Replace ---")
mystr = "MalaYa lAm"

# Convert to lowercase and strip out spaces
newstr = mystr.lower().replace(" ", "")
revstr = newstr[::-1]

print("Original:", mystr)
print("Cleaned :", newstr)
print("Reversed:", revstr)

if newstr == revstr:
    print("Is Palindrome:", True)
else:
    print("Is Palindrome:", False)
print()


# ---------------------------------------------------------------------
# Approach 3: Hardcoded Punctuation Removal
# ---------------------------------------------------------------------
print("--- Approach 3: Hardcoded Punctuation Removal ---")
mystr = "Sir, I demand, I am a maid named Iris"

# Chaining multiple replaces for specific symbols (spaces and commas)
newstr = mystr.lower().replace(" ", "").replace(",", "")
revstr = newstr[::-1]

print("Original:", mystr)
print("Cleaned :", newstr)
print("Reversed:", revstr)

if newstr == revstr:
    print("Is Palindrome:", True)
else:
    print("Is Palindrome:", False)
print()


# ---------------------------------------------------------------------
# Approach 4: Iterative Character Filtering (Universal Strategy)
# (Automatically drops spaces, dots, commas, and special characters)
# ---------------------------------------------------------------------
print("--- Approach 4: Iterative Character Filtering ---")
mystr = "Red roses run no risk. Sir, on Nurse's Order. "
newstr = ""

# Loop through every character to isolate alphabetic text only
for each in mystr:
    if each.isalpha():
        newstr += each.lower()

print("Original:", mystr)
print("Cleaned :", newstr)

if newstr == newstr[::-1]:
    print("Is Palindrome:", True)
else:
    print("Is Palindrome:", False)
print()


# ---------------------------------------------------------------------
# Approach 5: Ultimate Production Grade Reusable Function
# ---------------------------------------------------------------------
print("--- Approach 5: Palindrome Checker Function ---")

def Check_palindrome(mystr):
    """
    Cleans a string by extracting only alphanumeric characters 
    in lowercase, then checks if it reads the same forward and backward.
    """
    newstr = ""
    for each in mystr:
        if each.isalpha():
            newstr += each.lower()
            
    if newstr == newstr[::-1]:
        return True
    else:
        return False

# Running test cases against the function
print("Test 1 ('MalaYa lAm'):", Check_palindrome("MalaYa lAm"))
print("Test 2 ('racecar'):", Check_palindrome("racecar"))
print("Test 3 ('Red roses run no risk...'):", Check_palindrome("Red roses run no risk. Sir, on Nurse's Order. "))
print("Test 4 ('Sir, I demand...'):", Check_palindrome("Sir, I demand, I am a maid named Iris"))
```

```python

# =====================================================================
# PART 2: Character Frequency Mapping
# =====================================================================

# --- 6. Direct In-line Character Frequency Count ---
mystr5 = "HellO WorLD"
newstr5 = ''
for each in mystr5:
    if each.isalpha():
        newstr5 = newstr5 + each.lower()
print("\n--- Base Alpha Frequency String ---")
print(newstr5) # -> helloworld

mydict = {}
for each_1 in newstr5:
    if each_1 not in mydict.keys():
        mydict[each_1] = 1
    else:
        mydict[each_1] += 1
print(mydict)
# >>> OUTPUT:
# {'h': 1, 'e': 1, 'l': 3, 'o': 2, 'w': 1, 'r': 1, 'd': 1}

# --- 7. Reusable Frequency Map Function (.isalnum based) ---
def Char_Frequency(mystr):
    newstr = ''
    mydict = {}
    # Filter numbers & letters
    for each in mystr:
        if each.isalnum():
            newstr = newstr + each.lower()
    # Build dictionary map
    for each_1 in newstr:
        if each_1 not in mydict.keys():
            mydict[each_1] = 1
        else:
            mydict[each_1] += 1
    return mydict

print("\n--- Character Frequency Function Tests ---")
print(Char_Frequency("9840243168"))
# -> {'9': 1, '8': 2, '4': 2, '0': 1, '2': 1, '3': 1, '1': 1, '6': 1}
print(Char_Frequency("racecar"))
# -> {'r': 2, 'a': 2, 'c': 2, 'e': 1}
print(Char_Frequency("MalaYa lAm"))
# -> {'m': 2, 'a': 4, 'l': 2, 'y': 1}
print(Char_Frequency("Red roses run no risk. Sir, on Nurse's Order."))
# -> {'r': 8, 'e': 4, 'd': 2, 'o': 4, 's': 6, 'u': 2, 'n': 4, 'i': 2, 'k': 1}


# =====================================================================
# PART 3: Task - Remove Duplicate Values from List
# =====================================================================

# Note: In Python, `True == 1` and `False == 0`. To prevent boolean values 
# from being swallowed by numeric checks during standard loops, we verify exact types.
input_list = [10, 30.55, "hello", 30.55, 10, True, "Hello", "hello", False, 10, True]
output_list = []

for item in input_list:
    # Ensure item matches both value AND structural type before assuming it exists
    if not any(x == item and type(x) is type(item) for x in output_list):
        output_list.append(item)

print("\n--- Remove Duplicate Values Task ---")
print("Input :", input_list)
print("Output:", output_list)
# >>> OUTPUT:
# [10, 30.55, 'hello', True, 'Hello', False]
```
