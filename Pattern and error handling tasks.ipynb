**PRINT THE BELOW PATTERN:**
&&&&&
&&&&
&&&
&&
&

```python
rows = 5
for i in range(rows, 0, -1):
    for j in range(i):
        print("&", end="")
    print()
```

```python
# Print the below pattern (11/9/2026)
# -----
#  ----
#   ---
#    --
#     -

a = 5
spaces = 0
for i in range(a, 0, -1):
    for k in range(spaces):
        print(end=' ')
    spaces += 1
    for j in range(0, i):
        print("-",end="")
    print("\n",end="")
```

```python
# Print the below pattern (15/9/2026)
# *********
#  *******
#   *****
#    ***
#     *

a = 5
spaces = 0
for i in range(a, 0, -1):
    for k in range(spaces):
        print(end=' ')
    spaces += 1
    for j in range(0, 2*i-1):
        print("*",end="")
    print("\n",end="")
```
