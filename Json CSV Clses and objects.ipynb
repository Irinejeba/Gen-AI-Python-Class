```python
import json
import requests
from pprint import pprint

# --- SECTION 1: FETCHING LIVE JSON DATA FROM AN API ---
print("--- Fetching Astronaut Data ---")
url = "http://open-notify.org"
response = requests.get(url)

# Parse response into a Python dictionary
data_dict = response.json()

# 1. Print raw data
print("\n[Raw JSON Dictionary]:")
print(data_dict)

# 2. Pretty print full data
print("\n[Pretty Printed Data]:")
pprint(data_dict)

# 3. Pretty print just the 'people' list
print("\n[People Key Only]:")
pprint(data_dict['people'])

# 4. Access the first person in the list (Index 0)
print("\n[First Person in List]:")
pprint(data_dict['people'][0])

# 5. Loop through and print formatted details
print("\n[Formatted Astronaut List]:")
for each in data_dict['people']:
    print(f"{each['name']}\t=========\t{each['craft']}")


# --- SECTION 2: CONVERTING DICTIONARY TO JSON STRING ---
print("\n--- Dictionary to JSON String (dumps) ---")
mydict = {
    "Name": "Manjunathan",
    "City": ("chennai", "Mumbai", "Kovai"),
    "Year": None
}

print("Original Python Object:", mydict, type(mydict))

# Convert Python dict to JSON formatted string
json_string = json.dumps(mydict)
print("Converted JSON String: ", json_string, type(json_string))


# --- SECTION 3: WRITING JSON TO A FILE ---
print("\n--- Writing JSON File (dump) ---")
json_file_path = r"C:\Users\manju\Desktop\ajith.json"

with open(json_file_path, 'w') as myfile:
    json.dump(mydict, myfile)
print(f"Successfully saved JSON data to {json_file_path}")


# --- SECTION 4: READING JSON FROM A FILE ---
print("\n--- Reading JSON File (load) ---")
with open(json_file_path, 'r') as myfile:
    loaded_data = json.load(myfile)

print("Loaded Data From File:", loaded_data, type(loaded_data))
```

```python
# --- SECTION 1: BASIC CLASSES & OBJECTS ---
print("--- 1. Basic Class and Variable Reassignment ---")
class Vehicle:
    x = 10  # Class Variable
    y = 30  # Class Variable

    def myfunc(self):
        print("Hello My function Calling")
        print(f"Self object reference: {self}")

    def myfunc_with_args(self, a, b):  # a, b are function variables
        print("Hello My function Calling")
        print(f"The Function variables are = {a}, {b}")

    def myfunc_show_all(self, a, b):
        print("Hello My function Calling")
        print(f"The Function variables are = {a}, {b}")
        print(f"The Class Variables are = {self.x}, {self.y}")

# Testing object creation and variable override
obj1 = Vehicle()
obj2 = Vehicle()
print(f"Initial obj1: {obj1.x}, {obj1.y}")

obj1.x = "Venkat"  # Overriding instance variable x for obj1
print(f"Modified obj1: {obj1.x}, {obj1.y}")
print(f"Unmodified obj2: {obj2.x}, {obj2.y}")

# Testing functions with arguments
print("\n--- Functions with Arguments ---")
obj3 = Vehicle()
obj3.myfunc_with_args(100, 200)

# Testing accessing both class and local variables
print("\n--- Displaying Class vs Function Variables ---")
obj4 = Vehicle()
obj4.myfunc_show_all(40, 50)

# Modifying instance state and calling the function
obj5 = Vehicle()
obj5.x = "Manju"
obj5.myfunc_show_all(200, 400)


# --- SECTION 2: CONSTRUCTORS (__init__) ---
print("\n--- 2. Constructors (__init__) ---")
class XYZ:
    def __init__(self, x, y):
        print("Constructor calling")
        self.x = x  # Assigning parameter to instance variable
        self.y = y

obj_xyz1 = XYZ(100, 200)
print(f"Obj1 data: {obj_xyz1.x}, {obj_xyz1.y}")

obj_xyz2 = XYZ("Manju", 40)
print(f"Obj2 data: {obj_xyz2.x}, {obj_xyz2.y}")

# Student Class Example
class Student:
    def __init__(self, name, age, course):
        print("Student Data Created")
        self.x = name
        self.y = age
        self.z = course

venkat = Student("Venkat", 25, "Python")
print(f"Student 1: {venkat.x}, {venkat.y}, {venkat.z}")

baby = Student("Baby", 26, "GenAi")
print(f"Student 2: {baby.x}, {baby.y}, {baby.z}")


# --- SECTION 3: DESTRUCTORS (__del__) ---
print("\n--- 3. Destructors (__del__) ---")
class DestructorDemo:
    def __init__(self, name, age):
        print(f"Constructor calling for {name}")
        self.name = name
        self.age = age

    def __del__(self):
        print(f"Destructor executed for {self.name}")

# Creating objects to watch creation and cleanup sequence
obj_d1 = DestructorDemo("Honey", 30)
obj_d2 = DestructorDemo("Kani", 45)


# --- SECTION 4: POLYMORPHISM ---
print("\n--- 4. Polymorphism ---")
# Built-in function polymorphism
print(f"Length of string: {len('hello world')}")
print(f"Length of list: {len([20, 30.55, 'hello world'])}")

# User-defined function polymorphism
def add(a, b):
    return a + b

print(f"Addition (Numbers): {add(2, 3)}")
print(f"Addition (Strings): {add('Hi ', 'Manju')}")

# Class-based Polymorphism
class India:
    def capital(self):
        print("New Delhi is the capital of India.")
    def language(self):
        print("Hindi is the primary language of India.")
    def type(self):
        print("India is a developing country.")

class USA:
    def capital(self):
        print("Washington DC is the capital of USA.")
    def language(self):
        print("English is the primary language of USA.")
    def type(self):
        print("USA is a developed country.")

obj_ind = India()
obj_usa = USA()

print("\nExecuting polymorphic loop:")
for country in [obj_ind, obj_usa]:
    country.capital()
    country.language()
    country.type()


# --- SECTION 5: ENCAPSULATION (Private / Protected Members) ---
print("\n--- 5. Encapsulation ---")

class EncapsulationDemo:
    def __init__(self):
        self._a = 10  # Protected/Private variable by convention
        self._b = 20  # Protected/Private variable by convention

    def getattr(self):
        print(f"This is my Class variable data = {self._a}, {self._b}")

obj_encap = EncapsulationDemo()
obj_encap.getattr()

# Attempting external adjustment (Creates an external attribute instead of altering internal state logic)
obj_encap.a = 1000 
obj_encap.getattr()  # Notice the values inside remain unchanged

# Car Speed Encapsulation Example
class Car:
    def __init__(self):
        self.__maxspeed = 200  # Double underscore makes it strongly private

    def drive(self):
        print(f"driving. maxspeed: {self.__maxspeed}")

redcar = Car()
redcar.drive()

redcar.__maxspeed = 500  # Attempts to change it externally (fails silently by creating a new attribute)
redcar.drive()  # Will still safely display 200
```
