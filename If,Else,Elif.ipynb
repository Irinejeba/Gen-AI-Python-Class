```python
# ==========================================
# IF, ELIF, ELSE (DECISION STATEMENTS)
# ==========================================
# Rules:
# - 'if': Exact matching constraint. Only once per block (Mandatory).
# - 'elif': Handles alternative checks. Can be used as many times as needed (Optional).
# - 'else': Captures anything left over. Only once per block (Optional).

a = 100

if a > 100:
    print(a, "is bigger than 100")
elif a == 100:
    print(a, "is equal to 100")
else:
    print(a, "is smaller than 100")

print("This line Always works")  # Unindented line runs regardless of conditional checks


# Multiple 'elif' chain example
val_check = 25
if val_check > 50:
    print("val_check is bigger")
elif val_check == 20:
    print("val_check is 20")
elif val_check == 30:
    # Multiple branches allow granular conditional matching
    print("val_check is 30")  
elif val_check == 40:
    print("val_check is 40")
else:
    print("val_check is smaller")


# ==========================================
# NESTED IF & ELSE
# ==========================================
nested_val = 50

if nested_val > 10:
    print("nested_val is bigger")
    if nested_val > 100:
        print("nested_val IS BIGGEST")
    else:
        print("nested_val is between 10 to 99")
elif nested_val < 10:
    print("nested_val is smaller")
    if nested_val < 0:
        print("nested_val is negative")
    else:
        print("nested_val is between 0 to 9")
else:
    print("Code Error")

print("Code over")


# ==========================================
# SHORT HAND IF & ELSE (TERNARY OPERATORS)
# ==========================================
x = 1000
y = 500

# Inline structure Variant A (Shorthand if)
if x > y: print("x is bigger than y")
else: print("y is bigger than x")

# Inline structure Variant B (True value IF condition ELSE False value)
print("x is bigger than y") if x > y else print("y is bigger than x")


# ==========================================
# MINI GAME: GUESS THE SECRET NUMBER
# ==========================================
score = 100
secret_number = 50

# Type casting to 'int' is required since input() naturally captures text strings
guess = int(input("Enter your guess: "))

if guess == secret_number:
    print("Correct!")
    score = score + 20
elif guess > secret_number:
    print("Too High!")
    score = score - 10
else:
    print("Too Low!")
    score = score - 10

print("Your Score:", score)


# ==========================================
# BASIC TWO-NUMBER COMPARISON
# ==========================================
num1 = int(input("Enter a number A: "))
num2 = int(input("Enter a number B: "))

if num1 > num2:
    print("A is bigger than B")
elif num1 == num2:
    print("A is equal to B")
else:
    print("B is bigger than A")


# ==========================================
# SOLUTION FOR THE UNFINISHED TASK (PAGE 4)
# ==========================================
# Task: Find the largest among three user input numbers
print("\n--- Running Three-Number Maximum Task ---")
val_a = int(input("Enter a number A: "))
val_b = int(input("Enter a number B: "))
val_c = int(input("Enter a number C: "))

if val_a >= val_b and val_a >= val_c:
    print("A is the largest number.")
elif val_b >= val_a and val_b >= val_c:
    print("B is the largest number.")
else:
    print("C is the largest number.")
```
