**While Loop**

```python
# 26. 1 to 10
i=1
while i<=10:
    print(i,end=" "); i+=1
```

```python
# 27. 10 to 1
i=10
while i>=1:
    print(i,end=" "); i-=1
```

```python
# 28. Even 1 to 50
i=1
while i<=50:
    if i%2==0: print(i,end=" ") 
    i+=1
```

```python
# 29. Odd 1 to 50
i=1
while i<=50:
    if i%2!=0: print(i,end=" ")
    i+=1
```

```python
# 30. Sum 1 to 100
i=1; total=0
while i<=100:
    total+=i; i+=1
print(total)
```

```python
# 31. Multiplication table
num=int(input("Enter number: ")); i=1
while i<=10:
    print(f"{num} x {i} = {num*i}"); i+=1
```

```python
# 32. Factorial
num=int(input("Enter number: ")); fact=1; i=1
while i<=num:
    fact*=i; i+=1
print(fact)
```

```python
# 33. Count digits
num=int(input("Enter number: ")); count=0
while num>0:
    count+=1; num//=10
print(count)
```

```python
# 34. Reverse number
num=int(input("Enter number: ")); rev=0
while num>0:
    rev=rev*10+num%10; num//=10
print(rev)
```

```python
# 35. Palindrome
num=int(input("Enter number: ")); original=num; rev=0
while num>0:
    rev=rev*10+num%10; num//=10
print("Palindrome" if original==rev else "Not Palindrome")
```
