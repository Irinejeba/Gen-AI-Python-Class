```python
# --- SECTION 1: BASE CLOTHING CLASS (Data Abstraction & Encapsulation) ---
print("--- 1. Clothing Class Example ---")

class Clothing(object):
    def __init__(self, clothing_type, color, size, price=None):
        self.type = clothing_type
        self.color = color
        self.size = size
        self.price = price

    def set_price(self, price):
        self.price = price
        print(f"Setting the price of the {self.color} {self.type} to ${price}")

    def get_price(self):
        try:
            assert self.price is not None
            print(f"The {self.color} {self.type} costs ${self.price}.")
        except AssertionError:
            print(f"The price of the {self.color} {self.type} hasn't been set yet!")

    def promote(self, percentage):
        try:
            self.price = self.price * (1 - percentage / 100)
            print(f"The price of the {self.color} {self.type} has been reduced by {percentage}%.")
        except TypeError:
            print("Oops. Set an initial price first!")

# Testing the Clothing class
bluejeans = Clothing("jeans", "blue", 12)
bluejeans.get_price()
bluejeans.set_price(100)
bluejeans.get_price()
bluejeans.promote(25)
bluejeans.get_price()


# --- SECTION 2: SINGLE INHERITANCE ---
print("\n--- 2. Single Inheritance ---")

class A_Single:
    a = 100
    def myfunc(self):
        print("This is Myfunction Calling")
        print(f"This is A class variable: {self.a}")

class B_Single(A_Single):
    b = 200
    def newfunc(self):
        print("This is Newfunction Calling")
        print(f"This is B class variable: {self.b}")

obj_single = B_Single()
obj_single.myfunc()
print()
obj_single.newfunc()


# --- SECTION 3: MULTIPLE INHERITANCE ---
print("\n--- 3. Multiple Inheritance ---")

class A_Mult:
    a = 100
    def myfunc(self):
        print("This is Myfunction Calling")
        print(f"This is A class variable: {self.a}")

class B_Mult:
    b = 200
    def newfunc(self):
        print("This is Newfunction Calling")
        print(f"This is B class variable: {self.b}")

class C_Mult(A_Mult, B_Mult):
    c = 300
    def Oldfunc(self):
        print("This is Oldfunction Calling")
        print(f"This is C class variable: {self.c}")

obj_mult = C_Mult()
obj_mult.myfunc()
print()
obj_mult.newfunc()
print()
obj_mult.Oldfunc()


# --- SECTION 4: MULTILEVEL INHERITANCE ---
print("\n--- 4. Multilevel Inheritance ---")

class A_Level:
    a = 100
    def myfunc(self):
        print("This is Myfunction Calling")
        print(f"This is A class variable: {self.a}")

class B_Level(A_Level):
    b = 200
    def newfunc(self):
        print("This is Newfunction Calling")
        print(f"This is B class variable: {self.b}")

class C_Level(B_Level):
    c = 300
    def Oldfunc(self):
        print("This is Oldfunction Calling")
        print(f"This is C class variable: {self.c}")

obj_level = C_Level()
obj_level.myfunc()
print()
obj_level.newfunc()
print()
obj_level.Oldfunc()


# --- SECTION 5: HIERARCHICAL INHERITANCE ---
print("\n--- 5. Hierarchical Inheritance ---")

class A_Hier:
    a = 100
    def myfunc(self):
        print("This is Myfunction Calling")
        print(f"This is A class variable: {self.a}")

class B_Hier(A_Hier):
    b = 200
    def newfunc(self):
        print("This is Newfunction Calling")
        print(f"This is B class variable: {self.b}")

class C_Hier(A_Hier):
    c = 300
    def Oldfunc(self):
        print("This is Oldfunction Calling")
        print(f"This is C class variable: {self.c}")

print("[Testing Child Object 1 (B)]")
obj1_hier = B_Hier()
obj1_hier.myfunc()
print()
obj1_hier.newfunc()

print("\n[Testing Child Object 2 (C)]")
obj2_hier = C_Hier()
obj2_hier.myfunc()
print()
obj2_hier.Oldfunc()


# --- SECTION 6: HYBRID INHERITANCE (Diamond Problem Architecture) ---
print("\n--- 6. Hybrid Inheritance ---")

class A_Hyb:
    a = 100
    def myfunc(self):
        print("This is Myfunction Calling")
        print(f"This is A class variable: {self.a}")

class B_Hyb(A_Hyb):
    b = 200
    def newfunc(self):
        print("This is Newfunction Calling")
        print(f"This is B class variable: {self.b}")

class C_Hyb(A_Hyb):
    c = 300
    def Oldfunc(self):
        print("This is Oldfunction Calling")
        print(f"This is C class variable: {self.c}")

class D_Hyb(B_Hyb, C_Hyb):
    d = 400
    def yourfunc(self):
        print("This is Yourfunction Calling")
        print(f"This is D class variable: {self.d}")

obj_hybrid = D_Hyb()
obj_hybrid.myfunc()
print()
obj_hybrid.newfunc()
print()
obj_hybrid.Oldfunc()
print()
obj_hybrid.yourfunc()
```
