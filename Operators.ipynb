```python
# ==========================================
# ARITHMETIC OPERATORS
# ==========================================
# Symbols: +, -, *, /, **, //, %

print(10 + 3)   # Addition -> 13
print(10 - 3)   # Subtraction -> 7
print(10 * 3)   # Multiplication -> 30
print(10 / 3)   # Standard Division -> 3.3333333333333335
print(10 ** 3)  # Exponentiation (Power: 10 * 10 * 10) -> 1000
print(10 // 3)  # Floor Division (Quotient) -> 3
print(10 % 3)   # Modulus (Remainder) -> 1


# ==========================================
# ASSIGNMENT OPERATORS
# ==========================================
# Symbols: =, +=, -=, /=, *=, **=, //=, %=

a = 10
print(a)        # Output: 10

a += 5          # Equivalent to: a = a + 5
print(a)        # Output: 15

a *= 2          # Equivalent to: a = a * 2
print(a)        # Output: 30

b = 0
b += 20         # Equivalent to: b = b + 20
print(b)        # Output: 20
```

```python
# ==========================================
# PART 1: ADVANCED OPERATORS REFERENCE
# ==========================================

# 1. Comparison Operators (Evaluates to True or False)
# Symbols: ==, !=, >, <, >=, <=
print("--- Comparison Operators ---")
print(10 == 10)  # Equal to -> True
print(10 != 10)  # Not equal to -> False
print(10 > 10)   # Greater than -> False
print(10 >= 10)  # Greater than or equal to -> True
print(10 < 10)   # Less than -> False
print(10 <= 10)  # Less than or equal to -> True

# 2. Logical Operators
# and: Returns True if BOTH statements are true
# or:  Returns True if AT LEAST ONE statement is true
# not: Reverses the boolean result (True becomes False)
print('\n---- AND ----')
print(10 == 10 and 6 > 4)  # True and True -> True
print(10 != 10 and 6 > 4)  # False and True -> False
print(10 == 10 and 6 < 4)  # True and False -> False
print(10 != 10 and 6 < 4)  # False and False -> False

print('\n---- OR ----')
print(10 == 10 or 6 > 4)   # True or True -> True
print(10 != 10 or 6 > 4)   # False or True -> True
print(10 == 10 or 6 < 4)   # True or False -> True
print(10 != 10 or 6 < 4)   # False or False -> False

print('\n---- Not ----')
print(not 10 == 10)        # not True -> False
print(not 10 != 10)        # not False -> True


# ==========================================
# PART 2: CORE PYTHON COLLECTIONS
# ==========================================

# 1. LIST: Ordered, Changeable (Mutable), Allows Duplicates. Denoted by []
print("\n--- Collection: Lists ---")
mylist = [10, 20.55, "Hello", True, 10, "Hello", 20.55, False, True, 10, 20.55, True, 10, 10]
print(mylist, type(mylist))
print("Index 5 item:", mylist[5])  # Access item -> Hello
print("Index 8 item:", mylist[8])  # Access item -> True

mylist[8] = "Welcome"             # Lists are mutable! Modifying index 8
mylist.append("Chakaravarthi")    # Appending a new member to the list array
print("Updated list:", mylist)

# 2. TUPLE: Ordered, Unchangeable (Immutable), Allows Duplicates. Denoted by ()
print("\n--- Collection: Tuples ---")
mytuple = (10, 20.55, "Hello", True, 10, "Hello", 20.55, False, True, 10, 20.55, True, 10, 10)
print(mytuple, type(mytuple))
print("Index 4 item:", mytuple[4])  # Access item -> 10
print("Index 7 item:", mytuple[7])  # Access item -> False

# The line below is commented out because it will crash your script with a TypeError!
# Tuples are immutable and do not allow item assignment changes after creation.
# mytuple[7] = "Python" 

# 3. SET: Unordered, Unindexed, NO Duplicate Members. Denoted by {}
print("\n--- Collection: Sets ---")
# Duplicate literal values passed below will automatically be omitted upon conversion
myset = {100, "hai", 100, "hello", "hai", 50.55, 100, "hello", 50.55, True, False}
print(myset, type(myset))  # Output items show no sorting sequence order and no repeats!

# 4. DICTIONARY: Unordered, Changeable (Mutable), Key-Value Structure, Keys must be unique. Denoted by {k: v}
print("\n--- Collection: Dictionaries ---")
mydict = {
    "Name": "Jeevan",
    "Cours": "Python",
    "Age": 15,
    100: 50.55,
    40.55: True,
}
print(mydict, type(mydict))
print("Access value via key string ['Name']:", mydict["Name"])  # Output: Jeevan

mydict["Age"] = 16          # Updating value for an existing key
mydict["City"] = "Chennai"  # Creating a brand new key-value pair pair dynamically
print("Updated dictionary:", mydict)


# ==========================================
# PART 3: MUTABLE VS IMMUTABLE MEMORY BEHAVIOUR
# ==========================================

# IMMUTABLE OBJECTS (int, float, bool, str, tuple)
# - Identical scalar values map to the exact same memory address allocation (Optimization)
# - Changing the value forces creation of a separate distinct data allocation address
print("\n--- Memory Optimization: Immutable Objects ---")
var_a = 100
var_b = 100
var_c = 101
print(f"a: {var_a} | ID: {id(var_a)}")  # Identical memory location address map pointer
print(f"b: {var_b} | ID: {id(var_b)}")  # Identical address as var_a
print(f"c: {var_c} | ID: {id(var_c)}")  # Shifts to a brand new unique address location

tup_x = (10, 20, 30, 40, 50)
tup_y = (10, 20, 30, 40, 50)
print(f"Tuple x ID: {id(tup_x)}")
print(f"Tuple y ID: {id(tup_y)}")  # (Note: Literal tuples might evaluate matching addresses or vary across setups)

# MUTABLE OBJECTS (list, dict, set)
# - Identical data setups store inside entirely separate unique memory allocation addresses
print("\n--- Memory Isolation: Mutable Objects ---")
lst_x = [10, 20, 30, 40, 50]
lst_y = [10, 20, 30, 40, 50]
print(f"List x ID: {id(lst_x)}")  # Explicit unique allocation point A
print(f"List y ID: {id(lst_y)}")  # Explicit unique allocation point B (Different from x!)


# ==========================================
# PART 4: IDENTITY & MEMBERSHIP OPERATORS
# ==========================================

# 1. Identity Operators (is / is not) -> Checks memory LOCATION address equality (id)
print("\n--- Identity Check evaluations ---")
id_a = 100
id_b = 100
print("Value comparison (id_a != id_b):", id_a != id_b)        # Value checks match -> False
print("Location address check (id_a is not id_b):", id_a is not id_b) # Location address matches -> False

x_tuple = (10, 20, 30, 40, 50)
y_tuple = (10, 20, 30, 40, 50)
print("Tuple value mismatch check (x_tuple != y_tuple):", x_tuple != y_tuple) # False
print("Tuple location alignment check (x_tuple is y_tuple):", x_tuple is y_tuple)  # Evaluates position parity

# 2. Membership Operators (in / not in) -> Scans item arrays for existing target matches
print("\n--- Membership Check evaluations ---")
print("Substring 'e' search inside 'Hello world':", 'e' in "Hello world")  # True

parsed_list = [10, 20.55, "hello world"]
print("Item exact string matching validation check inside list array:", "hello world" not in parsed_list)  # False (it IS in the list)
```
