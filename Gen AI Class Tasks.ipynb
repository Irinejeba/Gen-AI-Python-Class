```python
score = 100
secret_number =50
guess = int(input("Enter your guess: "))
if guess == secret_number:  
    print("Correct")
    score = score + 20
    print ("Your score is:", score)

elif guess > secret_number:
    print("Your guess is too high. Try again.")
    score = score -10
    print("Your score is:", score)

else:
    print("too low.")
    score = score - 5  
    print("Your score is:", score)
```

```python
a = int(input("Enter a number :"))
b = int(input("Enter a number :"))
c = int(input("Enter a number :"))

if a > b and a > c:
    print("A is greatest")
elif b > a and b > c:
    print("B is greatest")
elif c > a and c > b:
    print("C is greatest")
else:
    print("Two or more numbers are equal and greatest")
```
