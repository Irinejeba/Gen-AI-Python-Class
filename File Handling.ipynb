```python
# ==========================================
# Python File Handling Examples
# ==========================================

# --- Section 1: Standard File Reading and Seeking ---
# Opening the file in default read mode
myfile = open("C:/Users/manju/Desktop/Mani.txt")

print(myfile.tell())         # Get current file position (0)
data = myfile.read(100)      # Read first 100 characters
print(data)
print(myfile.tell())         # Position after reading 100 characters

print('----------------')
data = myfile.read(200)      # Read next 200 characters
print(data)
print(myfile.tell())

myfile.seek(0)               # Move the cursor back to the beginning
print(myfile.tell())

print('----------------')
data = myfile.read(500)      # Read 500 characters from the start
print(data)
print(myfile.tell())

myfile.close()               # Always remember to close manual streams!


# --- Section 2: Reading with Context Managers ('with' keyword) ---
# Automatically handles closing the file context
with open("C:\\Users\\manju\\Desktop\\Mani.txt") as myfile:
    myline = myfile.readline()
    print(myline)


# --- Section 3: Reading Multiple Lines into Lists ---
with open("C:\\Users\\manju\\Desktop\\Mani.txt") as myfile:
    mylines = myfile.readlines()
    print(mylines)
    print('---------------------------------')
    print(mylines[2])       # Access the 3rd line specifically
    print('---------------------------------')
    for each in mylines:    # Loop and print each line
        print(each)


# --- Section 4: Writing & Modifying Files ---

# "a" - Append mode (Adds content to the end of the file)
with open("C:\\Users\\manju\\Desktop\\Mani.txt", 'a') as myfile:
    print(myfile.tell())
    mydata = myfile.write("\nWelcome All")

# "w" - Write mode (Overwrites existing file content completely)
with open("C:\\Users\\manju\\Desktop\\Mani.txt", 'w') as myfile:
    print(myfile.tell())
    mydata = myfile.write("Hello this GenAi class!!!!!!!!!")

# "x" - Create mode (Creates a file; throws FileExistsError if it already exists)
try:
    with open("C:\\Users\\manju\\Desktop\\vengat.txt", 'x') as myfile:
        mydata = myfile.write("Hello World")
except FileExistsError:
    print("Skipped: 'vengat.txt' already exists.")


# --- Section 5: Writing Structured Text lines ---
with open("C:\\Users\\manju\\Desktop\\Jeba.txt", 'w') as myfile:
    myfile.write("Name : Irine Jeba\n")
    myfile.write("Age : 25\n")
    myfile.write("Course : GenAi\n")
    myfile.write("City: Tambaram\n")
```
