```python
import pprint

# ==============================================================================
# 1. DICTIONARY CREATION & BASIC MANIPULATION
# ==============================================================================
print("--- 1. Dictionary Creation & Modification ---")
mydict = {
    "Name": "Honey",
    "Age": 35,
    "City": "Cennai",
    500: 50.55,
}

print(f"Original Dictionary: {mydict}")
print(f"Data Type: {type(mydict)}")
print(f"Accessing Key 500: {mydict[500]}\n")

# Modifying and adding new elements
mydict["City"] = "Mumbai"
mydict["Course"] = "Python"
print(f"Updated Dictionary: {mydict}\n\n")

# ==============================================================================
# 2. VIEW METHODS: keys(), values(), items()
# ==============================================================================
print("--- 2. View Methods ---")
# Resetting dictionary
mydict = {"Name": "Honey", "Age": 35, "City": "Cennai", 500: 50.55}

# keys() method
keys_list = list(mydict.keys())
print(f"Keys as List: {keys_list}")
print(f"Second Key in List: {keys_list[1]}")
print("Iterating over Keys:")
for key in mydict.keys():
    print(f" - {key}")
print()

# values() method
print(f"Values as List: {list(mydict.values())}")
print("Iterating over Values:")
for val in mydict.values():
    print(f" - {val}")
print()

# items() method
print(f"Items as List: {list(mydict.items())}")
print("Iterating over Items (Key & Value):")
for key, val in mydict.items():
    print(f" - {key}:\t{val}")
print("\n\n")

# ==============================================================================
# 3. REMOVAL METHODS: pop(), popitem(), clear()
# ==============================================================================
print("--- 3. Removal Methods ---")
# Set pop (Sets remove arbitrary elements)
y = {100, 20.55, "hai", 10}
popped_set_item = y.pop()
print(f"Set after pop(): {y} (Removed: {popped_set_item})")

# Dictionary pop(key)
mydict = {"Name": "Honey", "Age": 35, "City": "Cennai", 500: 50.55}
removed_val = mydict.pop("Age")
print(f"Dictionary after popping 'Age': {mydict} (Removed Value: {removed_val})")

# Dictionary popitem() (Removes last inserted item)
mydict = {"Name": "Honey", "Age": 35, "City": "Cennai", 500: 50.55}
mydict.popitem()
mydict.popitem()
mydict.popitem()
print(f"Dictionary after 3 popitem() calls: {mydict}")

# Dictionary clear()
mydict.clear()
print(f"Dictionary after clear(): {mydict}\n\n")

# ==============================================================================
# 4. MODIFICATION & FETCHING: update(), get()
# ==============================================================================
print("--- 4. Update & Fetching Methods ---")
mydict = {"Name": "Honey", "Age": 35, "City": "Cennai", 500: 50.55}
new_data = {"Name": "Manjunathan", "Age": 40, "City": "Kovai", "course": "Python"}

# Merging dictionaries
mydict.update(new_data)
print(f"Dictionary after update(): {mydict}\n")

# Re-assigning to original for .get() showcase
mydict = {"Name": "Honey", "Age": 35, "City": "Cennai", 500: 50.55}

# Safe key searching with .get() vs crashing with standard indexing []
print(f"Safe Fetch Existing Key (500): {mydict.get(500)}")
print(f"Safe Fetch Missing Key (600): {mydict.get(600)}")  # Returns None instead of crashing
print("Hello Python! (The script kept executing smoothly)\n\n")

# ==============================================================================
# 5. MEMORY COPYING: Direct Assignment vs copy()
# ==============================================================================
print("--- 5. Memory Copying Behaviors ---")
# Without .copy() (Both point to the same memory reference structure)
x = {"A": 100, "B": 200, "C": 300}
y = x
x["D"] = 400
y["E"] = 500
print("Shared reference (Modifying X impacts Y):")
print(f" X = {x}, ID = {id(x)}")
print(f" Y = {y}, ID = {id(y)}\n")

# With .copy() (Creates a shallow copy, unique memory location allocations)
x = {"A": 100, "B": 200, "C": 300}
y = x.copy()
x["D"] = 400
y["E"] = 500
print("Shallow copy (Modifying X does not impact Y):")
print(f" X = {x}, ID = {id(x)}")
print(f" Y = {y}, ID = {id(y)}\n\n")

# ==============================================================================
# 6. NESTED DICTIONARIES
# ==============================================================================
print("--- 6. Complex Nested Structures ---")
nested_dict = {
    "Name": "Manjunathan",
    "City": ["Chennai", "Kovai", "Mumbai"],
    "Year": (2000, 2010, 2020),
}

print(f"Accessing nested city index 0: {nested_dict.get('City')[0]}")
print(f"Accessing nested year index 1: {nested_dict.get('Year')[1]}")
print(f"Slicing nested string: {nested_dict.get('City')[2][3:]}\n")  # "Mumbai" -> "bai"

# JSON-like hierarchical menu block data parsing
menu_data = {
    "menu": {
        "id": "file",
        "value": "File",
        "popup": {
            "menuitem": [
                {"value": "New", "onclick": "CreateNewDoc()"},
                {"value": "Open", "onclick": "OpenDoc()"},
                {"value": "Close", "onclick": "CloseDoc()"},
            ]
        },
    }
}

print("Pretty printing complex structured hierarchy:")
pprint.pprint(menu_data)
print(f"\nTargeting deeper layered click actions:")
items = menu_data.get("menu")["popup"]["menuitem"]
print(f" Item 0 OnClick Action: {items[0]['onclick']}")
print(f" Item 2 OnClick Action: {items[2]['onclick']}")
```
