```python
# ==============================================================================
# --- PART 1: DECORATORS & FUNCTION CONCEPTS ---
# ==============================================================================

# 1. Nested Functions & Returning Functions
print("--- 1. Nested Function & Returning Functions ---")
def outer():
    n1 = 10
    def inner():
        n2 = 90
        n3 = n1 + n2
        return n3
    return inner

ans = outer()
print(f"Result of execution: {ans()}")
print(f"Function metadata name: {ans.__name__}\n")  # Fixed __name__ syntax


# 2. Closures (Functions that generate functions with presets)
print("--- 2. Closures (create_adder) ---")
def create_adder(x):
    def adder(y):
        return x + y
    return adder

# Numerical Addition Closure
num_adder = create_adder(100)
print(f"Numerical Addition (100 + 50): {num_adder(50)}")
print(f"Function metadata name: {num_adder.__name__}")

# String Concatenation Closure
string_adder = create_adder("Hello ")
print(string_adder("Honey"))
print(string_adder("Kani"))
print(string_adder("Mani\n"))


# 3. Functions as Parameters
print("--- 3. Passing Functions as Arguments ---")
def func1():
    print("Hi I'm function no 1")

def manju():
    print("Hello Manju")

def func2(x, y):
    print("Hi I'm function no 2")
    x()
    y()

func2(func1, manju)
print()


# 4. Working with Decorators (Modifying function behavior dynamically)
print("--- 4. Decorators (Without and With @ Syntax) ---")
def strupper(func):
    def inner():
        str1 = func()
        return str1.title()  # Changes string to Title Case as seen in page 3
    return inner

# Without using '@' syntax
def strname_raw():
    return "hi welcome"

def myname_raw():
    return "ManjuNaThan"

ans1 = strupper(strname_raw)
ans2 = strupper(myname_raw)
print(f"Manual Decorator 1: {ans1()}")
print(f"Manual Decorator 2: {ans2()}")

# With modern '@' decorator syntax
@strupper
def strname():
    return "hi welcome"

@strupper
def myname():
    return "ManjuNaThan"

print(f"Sugar Syntax Decorator 1: {strname()}")
print(f"Sugar Syntax Decorator 2: {myname()}\n")


# 5. Decorators handling Function Parameters & Safeguarding Errors
print("--- 5. Parameterized Decorators (Zero Division Guard) ---")
def divdec(func):
    def inner(x, y):
        if y == 0:
            return "Dont give divisor as zero"
        else:
            return func(x, y)
    return inner

@divdec
def div(a, b):
    return a / b

print(f"div(20, 0) -> {div(20, 0)}")
print(f"div(20, 3) -> {div(20, 3)}")
print(f"div(50, 2) -> {div(50, 2)}")
print(f"div(50, 0) -> {div(50, 0)}\n")


# ==============================================================================
# --- PART 2: TYPES OF METHODS IN CLASSES ---
# ==============================================================================

# 1. Instance Methods (Work with specific object states via 'self')
print("--- 6. Instance Methods ---")
class Subject:
    def __init__(self, tamil, english, maths):  # Fixed __init__ syntax
        self.tamil = tamil
        self.english = english
        self.maths = maths

    def marks(self):
        total = self.tamil + self.english + self.maths
        return total

kumar = Subject(90, 75, 100)
mani = Subject(70, 50, 60)
kavi = Subject(40, 35, 20)

print(f"Kumar's Total Marks: {kumar.marks()}")
print(f"Mani's Total Marks: {mani.marks()}")
print(f"Kavi's Total Marks: {kavi.marks()}\n")


# 2. Class Methods (Work with the class state via 'cls', instead of specific object data)
print("--- 7. Class Methods ---")
class DemoClassMethods:
    x = "hii a class method"

    @classmethod
    def method(cls, a, b):  # 'cls' is conventional for class methods
        return "The Total is " + str(a + b)

    @classmethod
    def method1(cls, x, y):
        print("The multiplication is " + str(x * y))
        return cls.x

obj_demo = DemoClassMethods()
# Calling from instances
print(obj_demo.method(300, 200))
print(obj_demo.method1(30, 2))
print('------------')
# Calling directly using Class reference
print(DemoClassMethods.method1(70, 3))
print(DemoClassMethods.method(40, 60))
print()


# 3. Static Methods (Standalone functions pinned to namespaces; no 'self' or 'cls')
print("--- 8. Static Methods ---")
class DemoStaticMethods:
    var = "Hi im a class method"

    def method(self, a, b):
        return "The total is " + str(a + b)

    @staticmethod
    def method1(x, y):
        print(x * y)
        return DemoStaticMethods.var

    @staticmethod
    def method2():
        return "hi am a static method"

obj_static = DemoStaticMethods()
print(obj_static.method(50, 40))
print(obj_static.method1(50, 4))
print(obj_static.method2())
```
