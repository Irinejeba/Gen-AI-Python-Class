```python
# Asci number
 
print("Manju")
print(ord("M"))   # Prints the ASCII value of 'M' -> 77  
print(ord("A"))   # Prints the ASCII value of 'A' -> 65
print(ord("N"))   # Prints the ASCII value of 'N' -> 78
print(ord("J"))   # Prints the ASCII value of 'J' -> 74
print(ord("U"))   # Prints the ASCII value of 'U' -> 85

print(chr(77))  # Prints the character corresponding to ASCII value 77 -> 'M'
print(chr(65))  # Prints the character corresponding to ASCII value 65 -> 'A'
print(chr(78))  # Prints the character corresponding to ASCII value 78 -> 'N'
print(chr(74))  # Prints the character corresponding to ASCII value 74 -> 'J'
print(chr(85))  # Prints the character corresponding to ASCII value 85 -> 'U'   

# ==========================================
# ADVANCED PRINT PARAMETERS (sep, end)
# ==========================================

# Using 'sep' to change the character between items (Default is a space)
print(1, 2, 3, 4, 5, sep="**")  # Output: 1**2**3**4**5
print(6, 7, 8, 9, 10)            # Output: 6 7 8 9 10

# Using 'end' to prevent jumping to a new line
print(1, 2, 3, 4, 5, end=" ")
print(6, 7, 8, 9, 10, end=" ")
print(11, 12, 13, 14, 15)       # All three prints appear on a single line


# ==========================================
# THE VALUE TRIO: value, type, id
# ==========================================

val_a = 100
print(val_a)        # The actual value -> 100
print(type(val_a))  # The structural data class -> <class 'int'>
print(id(val_a))    # Unique memory address location integer


# ==========================================
# TYPE CASTING (DATA TYPE CONVERSION)
# ==========================================

a = 500
a1 = float(a)    # Convert integer to float -> 500.0
a2 = str(a)      # Convert integer to string -> "500"
a3 = bool(a)     # Non-zero integer to boolean -> True

b = 40.55
b1 = int(b)      # Truncates decimal point -> 40
b2 = str(b)      # Convert float to string -> "40.55"
b3 = bool(b)     # Non-zero float to boolean -> True

c = "hello"
# c1 = int(c)    # ERROR: Strings with alphabetical characters cannot cast to integers
# c2 = float(c)  # ERROR: Strings with alphabetical characters cannot cast to floats
c3 = bool(c)     # Non-empty string evaluates to boolean -> True

d = "500"
d1 = int(d)      # Numerical string to integer -> 500
d2 = float(d)    # Numerical string to float -> 500.0
d3 = bool(d)     # Non-empty string evaluates to boolean -> True

# Special Boolean rules
e = bool("")     # Empty string -> False
e1 = bool(0)     # Numerical zero value -> False
print(e, type(e))
print(e1, type(e1))


# ==========================================
# DYNAMIC USER INPUT HANDLING
# ==========================================

# Standard input layout scenario A
print("Enter a Number :")
num = input()
print("The value of num = ", num)

# Cleaner layout scenario B using inside prompts
num = input("Enter a Number : ")
print("The value of num = ", num)

# Crucial Gotcha: input() captures everything as a string!
num1 = input("Enter a Number Num1 : ")  # e.g., enters 200
num2 = input("Enter a Number Num2 : ")  # e.g., enters 300
total = num1 + num2                     # Warning: Concatenates strings instead of adding!
print("The sum of (concatenated string result):", total)  # Outputs: 200300

# The proper solution: Explicit casting directly on string capture
num1_captured = input("Enter a Number Num1 : ")
x = int(num1_captured)  # Cast string explicitly to Integer
print("The value of num1 =", x, type(x))

num2_captured = float(input("Enter a Number Num2 : "))  # Cast directly inline to Float
print("The value of num2 =", num2_captured, type(num2_captured))

math_total = x + num2_captured
print("The authentic mathematical sum of:", math_total)


# ==========================================
# OBJECT DELETION MANAGEMENT (del keyword)
# ==========================================

temp_val = input("Enter a temporary Number: ")
x_converted = int(temp_val)

del temp_val  # Erases the reference label variable clean out of local memory

print("The value of X =", x_converted, type(x_converted))

# The line below is commented out because running it will intentionally crash your script!
# Python no longer recognizes this name setup after the 'del' instruction execution.
# print("The value of temp_val = ", temp_val)  # Throws a NameError Exception
```
