```python
# =====================================================================
# VS CODE PATTERN PROGRAMMING COMPENDIUM
# =====================================================================

def run_pattern_showcase(size=5):
    """
    Generates and prints a complete suite of standard matrix and loop patterns.
    Parameter 'size' controls the structural height/width of each pattern.
    """
    
    # -----------------------------------------------------------------
    # Pattern 1: Star Right-Angled Triangle (Nested Loop Method)
    # -----------------------------------------------------------------
    print("=== 1. STAR RIGHT-ANGLED TRIANGLE ===")
    for i in range(1, size + 1):
        for j in range(1, i + 1):
            print("*", end=" ")
        print()  # Terminal newline
    print()


    # -----------------------------------------------------------------
    # Pattern 2: Number Row Pyramid
    # -----------------------------------------------------------------
    print("=== 2. NUMBER ROW PYRAMID ===")
    for i in range(1, size + 1):
        for j in range(1, i + 1):
            print(i, end=" ")
        print()
    print()


    # -----------------------------------------------------------------
    # Pattern 3: Continuous Number Floyd's Triangle
    # -----------------------------------------------------------------
    print("=== 3. FLOYD'S TRIANGLE (CONTINUOUS NUMBERS) ===")
    count = 1
    for i in range(1, size + 1):
        for j in range(1, i + 1):
            print(f"{count:<2}", end=" ") # Padding formatting for grid alignment
            count += 1
        print()
    print()


    # -----------------------------------------------------------------
    # Pattern 4: Inverted Right-Angled Triangle
    # -----------------------------------------------------------------
    print("=== 4. INVERTED STAR TRIANGLE ===")
    for i in range(size, 0, -1):
        for j in range(1, i + 1):
            print("*", end=" ")
        print()
    print()


    # -----------------------------------------------------------------
    # Pattern 5: Symmetrical Full Equilateral Pyramid
    # -----------------------------------------------------------------
    print("=== 5. SYMMETRICAL FULL PYRAMID ===")
    for i in range(1, size + 1):
        # Spacing indentation shifts relative to remaining scale height
        print(" " * (size - i), end="")
        print("* " * i)
    print()


    # -----------------------------------------------------------------
    # Pattern 6: Symmetrical Diamond Shape
    # -----------------------------------------------------------------
    print("=== 6. SYMMETRICAL DIAMOND PATTERN ===")
    # Top Half of Diamond
    for i in range(1, size + 1):
        print(" " * (size - i), end="")
        print("* " * i)
    # Bottom Half of Diamond
    for i in range(size - 1, 0, -1):
        print(" " * (size - i), end="")
        print("* " * i)
    print()


    # -----------------------------------------------------------------
    # Pattern 7: Alphabet Character Increment Matrix
    # -----------------------------------------------------------------
    print("=== 7. ALPHABET INCREMENT TRIANGLE ===")
    # ASCII 65 corresponds to the character 'A'
    base_ascii = 65 
    for i in range(0, size):
        current_char = chr(base_ascii + i)
        for j in range(0, i + 1):
            print(current_char, end=" ")
        print()
    print()


# =====================================================================
# SCRIPT EXECUTION BLOCK
# =====================================================================
if __name__ == "__main__":
    # Change this variable to dynamically resize all outputs below!
    TARGET_SIZE = 5 
    
    print(f"Launching pattern generator with dimensions: {TARGET_SIZE}x{TARGET_SIZE}\n")
    run_pattern_showcase(TARGET_SIZE)

```
